```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import pearsonr, linregress
from sklearn.linear_model import LinearRegression
from sklearn.metrics import mean_squared_error
```

```python
df1= pd.read_csv("/content/sample_data/grupo1.csv")
df1.head()

print("Primeras filas del grupo 1:")
print(df1.head())
print()
print("Descripcion estadistica del grupo 1:")
print(df1.describe())
```

```python
plt.figure(figsize=(12, 5))

plt.subplot(1, 2, 1)
plt.boxplot(df1["HorasEstudio"], patch_artist=True, boxprops=dict(facecolor='orange'))
plt.title("Horas de estudio - Grupo 1")
plt.ylabel("Horas")
plt.grid(True, axis='y')

plt.subplot(1, 2, 2)
plt.boxplot(df1["Calificacion"], patch_artist=True, boxprops=dict(facecolor='skyblue'))
plt.title("Calificacion final - Grupo 1")
plt.ylabel("Calificacion")
plt.grid(True, axis='y')

plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df1["HorasEstudio"], df1["Calificacion"], color='skyblue', edgecolor='black')
plt.title("Grupo 1: Horas de estudio vs Calificacion")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificacion final")
plt.grid(True)
plt.show()
```

```python
plt.figure(figsize=(7, 4))
plt.hist(df1["Calificacion"], bins=10, color='skyblue', edgecolor='black')
plt.title("Histograma de calificaciones - Grupo 1")
plt.xlabel("Calificacion final")
plt.ylabel("Frecuencia")
plt.grid(True, axis='y')
plt.show()
```

```python
x1 = df1[["HorasEstudio"]]
y1 = df1["Calificacion"]

modelo1 = LinearRegression()
modelo1.fit(x1, y1)

pendiente1 = modelo1.coef_[0]
intercepto1 = modelo1.intercept_

print(f"Pendiente: {pendiente1:.4f}")
print(f"Intercepto: {intercepto1:.4f}")
print(f"Ecuacion: Calificacion = {intercepto1:.4f} + {pendiente1:.4f} * Horas")
```

```python
r1, p_corr1 = pearsonr(df1["HorasEstudio"], df1["Calificacion"])

print(f"Coeficiente de correlacion de Pearson (r): {r1:.4f}")
print(f"Coeficiente de determinacion (R2): {r1 ** 2:.4f}")
```

```python
reg1 = linregress(df1["HorasEstudio"], df1["Calificacion"])
print(f"Valor p: {reg1.pvalue:.6e}")
```

```python
plt.figure(figsize=(8, 5))
plt.scatter(df1["HorasEstudio"], df1["Calificacion"], color='skyblue', edgecolor='black', label="Estudiantes")
plt.plot(df1["HorasEstudio"], modelo1.predict(x1), color='red', linewidth=2, label="Linea de regresion")
plt.title(f"Grupo 1: Horas de estudio vs Calificacion (r = {r1:.4f})")
plt.xlabel("Horas de estudio")
plt.ylabel("Calificacion final")
plt.legend()
plt.grid(True)
plt.show()
```

##1.   ¿Explicar cuál es la interpretación del coeficiente de regresión en términos de la relación entre el tiempo de estudio y las calificaciones finales?

*   La pendiente es de −0.106, es decir, prácticamente nula. Estudiar una hora adicional no modifica de manera apreciable la calificación. Como su intervalo de confianza contiene el cero, ni siquiera es posible afirmar que la tendencia sea negativa. El intercepto (85.3) queda muy cerca del promedio del grupo (83.8).

Grupo 2: Por cada hora adicional de estudio, la calificación aumenta en promedio unos 4.2 puntos, con un intervalo de confianza de 3.5 a 4.9 puntos.

Grupo 3: Cada hora extra se asocia con un aumento de alrededor de 5.3 puntos, con un intervalo de 4.7 a 5.8. El intercepto (≈ 0.1) es coherente: sin horas de estudio, la calificación esperada es prácticamente cero.

##2.   ¿Cómo se interpreta el coeficiente de correlación en términos de la fuerza y dirección de la relación entre las variables?

*   Grupo 1 (r = −0.057): No hay relación. Los puntos se reparten en una nube sin patrón, y el tiempo de estudio explica apenas el 0.3% de la variabilidad de las calificaciones.

Grupo 2 (r = 0.862): La relación es positiva y fuerte. A más horas de estudio, mejores calificaciones, y el modelo explica el 74.3% de la variabilidad (R²).

Grupo 3 (r = 0.942): La relación es positiva y muy fuerte, con un 88.8% de la variabilidad explicada. Sus puntos son los que más se pegan a la recta.


#3.   ¿El modelo de regresión lineal es estadísticamente significativo?
Con un nivel de significancia de 0.05:

Grupo 1: No es significativo. Con p = 0.69 no hay evidencia de una relación lineal entre horas y calificación.
Grupo 2: Sí es significativo. Con p ≈ 9×10⁻¹⁶ se descarta que la pendiente sea cero.
Grupo 3: Sí es significativo, con evidencia aún más contundente (p ≈ 2×10⁻²⁴).

#4.    ¿Qué tan confiable es el modelo de regresión para predecir las calificaciones finales de los estudiantes basado en el tiempo de estudio?

*   Grupo 1: No es confiable. Usar el modelo equivale a predecir “cerca de 84” para cualquier estudiante. El error típico (7.7 puntos) es casi igual a la desviación estándar de las calificaciones (7.8), por lo que el modelo no aporta mejora alguna.

Grupo 2: Su confiabilidad es moderada. Explica una parte importante de la variación, pero sus predicciones fallan en promedio unos 10 puntos, un margen considerable en una escala de 0 a 100. Además, los datos se dispersan más con pocas y con muchas horas de estudio.

Grupo 3: Es el más confiable de los tres (R² = 0.89, error típico de unos 8 puntos), aunque presenta dos limitaciones:

La relación no es del todo lineal. Los residuos son negativos con pocas y con muchas horas, y positivos en la zona media, lo que revela una curva que se aplana cerca de 100. Un ajuste cuadrático elevaría el R² a ≈ 0.95. En el grupo 2 ocurre algo similar, aunque con menor intensidad.
Existe un límite de 100 puntos. Para 20 horas la recta predice 105.8, un valor imposible. Por ello, el modelo no debe usarse fuera del rango observado (≈ 4 a 20 horas).

En todos los grupos la muestra es de solo 50 estudiantes, y una correlación alta no basta para demostrar que estudiar cause las calificaciones.
