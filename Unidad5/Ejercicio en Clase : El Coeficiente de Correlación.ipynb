```python
from scipy.stats import pearsonr
```

```python
# Datos de Entrada
publicidad = [2, 3, 5, 4, 11, 5]
ganancias = [20 ,25, 34, 30, 40, 31]

#Calcular el coeficiente de de correlación de Pearson
coefficient_correlation, p_valor = pearsonr(publicidad, ganancias)

print('El Coeficiente de Correlación de Pearson es: ', coefficient_correlation)
print('Valor p: ', p_valor)
```
